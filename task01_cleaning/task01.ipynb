#### LÀM SẠCH menu.csv

```python
import numpy as np

# Chuẩn hóa product_id: ép kiểu chuỗi và loại bỏ khoảng trắng thừa
menu['product_id'] = (
    menu['product_id']
    .astype(str)
    .str.strip()
)

# Chuẩn hóa product_name: ép kiểu chuỗi, bỏ khoảng trắng, viết hoa chữ cái đầu
menu['product_name'] = (
    menu['product_name']
    .astype(str)
    .str.strip()
    .str.title()
)

# Chuẩn hóa price:
# - Ép sang chuỗi
# - Loại bỏ ký tự không phải số hoặc dấu .
# - Chuyển chuỗi rỗng thành NaN
# - Ép về kiểu float
menu['price'] = (
    menu['price']
    .astype(str)
    .str.replace(r'[^0-9.]', '', regex=True)
    .replace('', np.nan)
    .astype(float)
)

print("MENU SAU KHI LÀM SẠCH:")
display(menu.head())
```
