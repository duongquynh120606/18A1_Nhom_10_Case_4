#### TASK 01 – ĐỌC & LÀM SẠCH DỮ LIỆU

ĐỌC DỮ LIỆU GỐC

```python
import pandas as pd

menu = pd.read_csv('../data_raw/menu.csv')
order = pd.read_csv('../data_raw/order_log.csv')
cost = pd.read_csv('../data_raw/cost_raw_material.csv')
print("Đã đọc dữ liệu từ các file CSV")
```

#### LÀM SẠCH menu.csv

```python
import numpy as np

# Chuẩn hóa product_id: ép kiểu chuỗi và loại bỏ khoảng trắng thừa
menu['product_id'] = (
    menu['product_id']
    .astype(str)
    .str.strip()
)

# Chuẩn hóa product_name: ép kiểu chuỗi, bỏ khoảng trắng, viết hoa chữ cái đầu
menu['product_name'] = (
    menu['product_name']
    .astype(str)
    .str.strip()
    .str.title()
)

# Chuẩn hóa price:
# - Ép sang chuỗi
# - Loại bỏ ký tự không phải số hoặc dấu .
# - Chuyển chuỗi rỗng thành NaN
# - Ép về kiểu float
menu['price'] = (
    menu['price']
    .astype(str)
    .str.replace(r'[^0-9.]', '', regex=True)
    .replace('', np.nan)
    .astype(float)
)

print("MENU SAU KHI LÀM SẠCH:")
display(menu.head())
```

#### LÀM SẠCH order_log.csv

```python
# Chuẩn hóa product_id: ép kiểu chuỗi và loại bỏ khoảng trắng
order['product_id'] = (
    order['product_id']
    .astype(str)
    .str.strip()
)

# Chuẩn hóa qty:
# - Chuyển về dạng số
# - Giá trị sai định dạng sẽ thành NaN
order['qty'] = pd.to_numeric(
    order['qty'],
    errors='coerce'
)

# Loại bỏ các dòng có qty không hợp lệ (NaN hoặc <= 0)
order = order[order['qty'] > 0]

# Chuẩn hóa order_time:
# - Đưa về kiểu datetime
# - Giá trị sai định dạng sẽ thành NaT
order['order_time'] = pd.to_datetime(
    order['order_time'],
    errors='coerce'
)

print("ORDER SAU KHI LÀM SẠCH:")
display(order.head())
```

#### LÀM SẠCH cost_raw_material.csv

```python
# Chuẩn hóa product_id:
# - Ép kiểu chuỗi
# - Loại bỏ khoảng trắng thừa
cost['product_id'] = (
    cost['product_id']
    .astype(str)
    .str.strip()
)

# Chuẩn hóa unit_cost:
# - Chuyển về dạng số
# - Giá trị sai định dạng sẽ thành NaN
cost['unit_cost'] = pd.to_numeric(
    cost['unit_cost'],
    errors='coerce'
)

print("COST SAU KHI LÀM SẠCH:")
display(cost.head())
```
