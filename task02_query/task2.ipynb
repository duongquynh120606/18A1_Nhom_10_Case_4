### TASK 02 – TRUY VẤN & THỐNG KÊ MÔ TẢ

#### IMPORT & LOAD DỮ LIỆU SẠCH

```python
import pandas as pd
import numpy as np

menu = pd.read_csv('../data_clean/menu_clean.csv')
order = pd.read_csv('../data_clean/order_log_clean.csv')
cost = pd.read_csv('../data_clean/cost_raw_material_clean.csv')

print("Menu:")
display(menu.head())

print("Order:")
display(order.head())

print("COST:")
display(cost.head())
```

#### XỬ LÝ THỜI GIAN ORDER (NGÀY – GIỜ)

```python
order['order_time'] = pd.to_datetime(order['order_time'], errors='coerce')

order['hour'] = order['order_time'].dt.hour
order['date'] = order['order_time'].dt.date

print("ORDER SAU KHI TÁCH NGÀY & GIỜ:")
display(order[['order_time', 'hour', 'date']].head())
```

#### THỐNG KÊ SỐ LƯỢNG MÓN BÁN THEO GIỜ

```python


sales_by_hour = (
    order
    .groupby('hour')['qty']
    .sum()
    .reset_index(name='total_qty')
    .sort_values('hour')
)

print("SỐ MÓN BÁN THEO KHUNG GIỜ:")
display(sales_by_hour)
```

#### TÌM MÓN BÁN CHẠY NHẤT

```python

top_product = (
    order
    .groupby('product_id')['qty']
    .sum()
    .reset_index()
    .sort_values('qty', ascending=False)
    .head(1)
)

print("MÓN BÁN CHẠY NHẤT:")
display(top_product)
```

#### TÍNH DOANH THU THEO TỪNG MÓN

```python


order_menu = order.merge(
    menu[['product_id', 'product_name', 'price']],
    on='product_id',
    how='left'
)

order_menu['revenue'] = order_menu['qty'] * order_menu['price']

revenue_by_product = (
    order_menu
    .groupby(['product_id', 'product_name'])['revenue']
    .sum()
    .reset_index()
    .sort_values('revenue', ascending=False)
)

print("DOANH THU THEO TỪNG MÓN:")
display(revenue_by_product.head())
```
