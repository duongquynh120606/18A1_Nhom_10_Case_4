```python
# CELL 1 – IMPORT & ĐỌC DỮ LIỆU

import pandas as pd
import numpy as np
import os

df = pd.read_csv("../data_merge/order_full_merged.csv")

print("DỮ LIỆU ĐẦU VÀO:")
display(df.head())
```

```python
# CELL 2 – DOANH THU THEO SẢN PHẨM

revenue_by_product = (
    df.groupby(["product_id", "product_name"])
      .agg(
          total_qty=("qty", "sum"),
          total_revenue=("revenue", "sum")
      )
      .reset_index()
      .sort_values("total_revenue", ascending=False)
)

print("DOANH THU THEO SẢN PHẨM:")
display(revenue_by_product)
```

```python
# CELL 3 – LỢI NHUẬN THEO SẢN PHẨM

profit_by_product = (
    df.groupby(["product_id", "product_name"])
      .agg(
          avg_profit=("profit", "mean"),
          unit_cost=("unit_cost", "mean"),
          price=("price", "mean")
      )
      .reset_index()
      .sort_values("avg_profit", ascending=False)
)

print("LỢI NHUẬN TRUNG BÌNH THEO SẢN PHẨM:")
display(profit_by_product)
```
