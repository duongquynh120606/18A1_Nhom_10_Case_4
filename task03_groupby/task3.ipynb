### GroupBy & Tổng hợp

#### IMPORT THƯ VIỆN & ĐỌC DỮ LIỆU

```python
import pandas as pd
import numpy as np

menu = pd.read_csv("../data_clean/menu_clean.csv")
order = pd.read_csv("../data_clean/order_log_clean.csv")
cost = pd.read_csv("../data_clean/cost_raw_material_clean.csv")

print("MENU:")
display(menu.head())

print("ORDER:")
display(order.head())

print("COST:")
display(cost.head())
```

#### MERGE ORDER VỚI MENU

```python
order_menu = order.merge(
    menu[['product_id', 'product_name', 'price']],
    on='product_id',
    how='left'
)

order_menu['revenue'] = order_menu['qty'] * order_menu['price']

print("ORDER SAU KHI MERGE MENU:")
display(order_menu.head())
```

#### MERGE THÊM COST & TÍNH LỢI NHUẬN

```python
order_full = order_menu.merge(
    cost[['product_id', 'unit_cost']],
    on='product_id',
    how='left'
)

order_full['profit'] = order_full['price'] - order_full['unit_cost']

print("ORDER SAU KHI GỘP COST & TÍNH LỢI NHUẬN:")
display(order_full.head())
```
