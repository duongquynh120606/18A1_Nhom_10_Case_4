### GroupBy & Tổng hợp

#### IMPORT THƯ VIỆN & ĐỌC DỮ LIỆU

```python
import pandas as pd
import numpy as np

menu = pd.read_csv("../data_clean/menu_clean.csv")
order = pd.read_csv("../data_clean/order_log_clean.csv")
cost = pd.read_csv("../data_clean/cost_raw_material_clean.csv")

print("MENU:")
display(menu.head())

print("ORDER:")
display(order.head())

print("COST:")
display(cost.head())
```

#### MERGE ORDER VỚI MENU

```python
order_menu = order.merge(
    menu[['product_id', 'product_name', 'price']],
    on='product_id',
    how='left'
)

order_menu['revenue'] = order_menu['qty'] * order_menu['price']

print("ORDER SAU KHI MERGE MENU:")
display(order_menu.head())
```

#### MERGE THÊM COST & TÍNH LỢI NHUẬN

```python
order_full = order_menu.merge(
    cost[['product_id', 'unit_cost']],
    on='product_id',
    how='left'
)

order_full['profit'] = order_full['price'] - order_full['unit_cost']

print("ORDER SAU KHI GỘP COST & TÍNH LỢI NHUẬN:")
display(order_full.head())
```

#### TÍNH DOANH THU THEO TỪNG MÓN

```python
revenue_by_product = (
    order_full
    .groupby(['product_id', 'product_name'])['revenue']
    .sum()
    .reset_index()
    .sort_values('revenue', ascending=False)
)

print("DOANH THU THEO TỪNG MÓN:")
display(revenue_by_product)
```

#### TÍNH LỢI NHUẬN TRUNG BÌNH THEO MÓN

```python
profit_by_product = (
    order_full
    .groupby(['product_id', 'product_name'])['profit']
    .mean()
    .reset_index()
    .sort_values('profit', ascending=False)
)

print("LỢI NHUẬN TRUNG BÌNH THEO MÓN:")
display(profit_by_product)
```

#### XÁC ĐỊNH MÓN CÓ BIÊN LỢI NHUẬN CAO NHẤT

```python
top_profit_product = profit_by_product.head(1)

print("MÓN CÓ BIÊN LỢI NHUẬN CAO NHẤT:")
display(top_profit_product)
```

#### DOANH THU THEO NGÀY VÀ GIỜ

```python
order_full['order_time'] = pd.to_datetime(order_full['order_time'], errors='coerce')
order_full['date'] = order_full['order_time'].dt.date
order_full['hour'] = order_full['order_time'].dt.hour

revenue_by_date_hour = (
    order_full
    .groupby(['date', 'hour'])['revenue']
    .sum()
    .reset_index()
)

print("DOANH THU THEO NGÀY & GIỜ:")
display(revenue_by_date_hour.head())
```

#### TÌM MÓN CÓ DOANH THU THẤP NHẤT

```python
low_performance_product = revenue_by_product.tail(1)

print("MÓN CÓ HIỆU SUẤT THẤP NHẤT:")
display(low_performance_product)
```
